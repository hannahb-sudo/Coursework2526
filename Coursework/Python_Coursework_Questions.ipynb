### Question 6
Using the startwars dataset imported below, identify the individuals with tallest height and lowest mass.

````python
starwars_dataset="https://www.fabricionarcizo.com/post/starwars/updated_starwars.csv"
starwars = pd.read_csv(starwars_dataset)
````

````python
#Tallest individual in the dataset
sw_tallest = max(starwars["height"])
print(sw_tallest) #Max value for height
sw_height = starwars.loc[starwars["height"] > 260, "name"]
print(sw_height) #Name of tallest individual

#Individual with the lowest mass
sw_lightest = min(starwars["mass"])
print(sw_lightest) #Min value for mass
sw_mass = starwars.loc[starwars["mass"] < 17, "name"]
print(sw_mass)
````

### Question 7
Analyze the life expectancy dataset (which can be loaded by running the cell below).

````python
import seaborn as sns
life_expectancy = sns.load_dataset('healthexp')
life_expectancy
````

````python
#Unique number of countries
le_country = life_expectancy["Country"].unique()
print("Countries:")
print(le_country)

#Country and year with highest life expectancy
le_highest = life_expectancy["Life_Expectancy"].max()
life_exp = life_expectancy[life_expectancy["Life_Expectancy"] > 84.5] #Select row with highest life expectancy
print(life_exp)

#Year with the highest total expenditure
le_group = life_expectancy.groupby("Year")["Spending_USD"].sum() #total expenditure per year
le_exp = le_group.max() #Highest total expenditure
print(le_exp)
````

### Question 8
Using the diamonds dataset and the plotting code of your choice, plot the following from the diamonds dataset.

1. Carat vs Price in a scatter plot.
2. A histogram of prices.
3. A boxplot of cut vs price.

The code can be loaded from seaborn in the cell below.

````python
diamonds = sns.load_dataset('diamonds')
diamonds
````

````python
#Scatterplot of carat vs. price
plt.scatter(diamonds['carat'], diamonds['price'], color='pink')
plt.xlabel('Carat')
plt.ylabel('Price')
plt.title('Diamonds Carat vs. Price')
plt.show()

#Histogram of diamond prices
plt.hist(diamonds['price'], bins=15, color='hotpink')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Histogram of Diamond Price')
plt.show()

#Boxplot of cut vs. price
sns.boxplot(x='cut', y='price', data=diamonds, color='pink')
plt.title('Diamonds Cut vs. Price')
plt.show()


````

### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.

1 carat = 0.2 grams

````python
#function for carat -> grams
diamonds["grams"] = diamonds.apply(lambda row: row.carat * 20, axis = 1)

#Heaviest diamond
heaviest_diamond = diamonds["grams"].max()
print(heaviest_diamond, "g")
````
